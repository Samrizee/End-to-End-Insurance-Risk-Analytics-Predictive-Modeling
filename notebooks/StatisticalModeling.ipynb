```python
import pandas as pd
```

```python
data = pd.read_csv('/home/samrawit/End-to-End-Insurance-Risk-Analytics-Predictive-Modeling/data/processed/MLRnew.csv')
```

```python
missing_values = data.isnull().sum()
print("Missing Values:\n", missing_values)
```

Impute the missing values

```python
# Impute missing values for categorical columns with the mode
categorical_cols = ['Bank', 'AccountType', 'Gender', 'MaritalStatus']
for col in categorical_cols:
    data[col].fillna(data[col].mode()[0], inplace=True)

```

```python
missing_values = data.isnull().sum()
print("Missing Values:\n", missing_values)
```

```python
# Remove columns with more than 70% missing values
threshold = 0.7 * len(data)
data = data.dropna(thresh=threshold, axis=1)
```

```python
missing_values = data.isnull().sum()
print("Missing Values:\n", missing_values)
```
