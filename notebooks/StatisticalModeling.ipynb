```python
import pandas as pd
```

```python
data = pd.read_csv('/home/samrawit/End-to-End-Insurance-Risk-Analytics-Predictive-Modeling/data/processed/MLRnew.csv')
```

```python
missing_values = data.isnull().sum()
print("Missing Values:\n", missing_values)
```

Impute the missing values

```python
# Impute missing values for categorical columns with the mode
categorical_cols = ['Bank', 'AccountType', 'Gender', 'MaritalStatus']
for col in categorical_cols:
    data[col].fillna(data[col].mode()[0], inplace=True)

```

```python
missing_values = data.isnull().sum()
print("Missing Values:\n", missing_values)
```

## Feature enginerring

```python
claim_data = data[data['TotalClaims'] > 0]
```

```python
claim_data['ClaimToPremiumRatio'] = claim_data['TotalClaims'] / claim_data['TotalPremium']
claim_data['PolicyAge'] = 2023 - claim_data['RegistrationYear']
claim_data['VehicleIntroDate'] = pd.to_datetime(claim_data['VehicleIntroDate'], errors='coerce')
claim_data['HighValueVehicle'] = (claim_data['CustomValueEstimate'] > 50000).astype(int)
claim_data['PerformanceIndex'] = claim_data['kilowatts'] / claim_data['cubiccapacity']
```
