### Claim Frequency

```python
# Create a new column for Claim Frequency
data['ClaimFrequency'] = np.where(data['TotalClaims'] > 0, 1, 0)
```

```python
# Calculate the overall claim frequency
overall_claim_frequency = data['ClaimFrequency'].mean()
print("Overall Claim Frequency:", overall_claim_frequency)
```

## Claim Severity 

```python
# Create a column to indicate if a claim has been made
data['HasClaim'] = data['TotalClaims'] > 0

# Calculate Claim Severity only for those who made a claim
data['ClaimSeverity'] = np.where(data['HasClaim'], data['TotalClaims'] / data['TotalPremium'], 0)

# To get the average claim severity for those who made a claim
average_claim_severity = data[data['HasClaim']]['ClaimSeverity'].mean()
print("Average Claim Severity:", average_claim_severity)
```

The output of Average Claim Severity: inf indicates that there is a division by zero occurring in the calculation of ClaimSeverity. This typically happens when TotalPremium is zero for some records where TotalClaims is greater than zero.

```python
zero_premium_count = data[data['TotalPremium'] == 0].shape[0]
print("Number of records with zero TotalPremium:", zero_premium_count)
```

```python

# Mark who made a claim
data['HasClaim'] = data['TotalClaims'] > 0

# Calculate claim severity ONLY for rows where a claim occurred
data['ClaimSeverity'] = np.where(
    data['HasClaim'],
    data['TotalClaims'],  # Severity = total claim amount (assuming it's the total for that policy)
    np.nan                # No claim → severity is undefined
)

# Average only for those who made a claim
average_claim_severity = data.loc[data['HasClaim'], 'ClaimSeverity'].mean()

print("Average Claim Severity:", average_claim_severity)
```

### Statistical Testing

#### Risk Differences Across Provinces (Claim Frequency)

```python

data['HasClaim'] = data['TotalClaims'] > 0
contingency_table_prov = pd.crosstab(data['Province'], data['HasClaim'])


chi2_prov, p_prov, dof, expected = stats.chi2_contingency(contingency_table_prov)

print("Chi-squared Test p-value for risk across Provinces:", p_prov)

if p_prov < 0.05:
    print("Reject H₀: There are risk differences across provinces.")
else:
    print("Accept H₀: There are no risk differences across provinces.")
```

Risk Differences Between Women and Men

```python
contingency_table_gender = pd.crosstab(data['Gender'], data['HasClaim'])

chi2_gender, p_gender, dof, expected = stats.chi2_contingency(contingency_table_gender)

print("Chi-squared Test p-value for risk between Women and Men:", p_gender)

if p_gender < 0.05:
    print("Reject H₀: There are significant risk differences between Women and Men.")
else:
    print("Accept H₀: There are no significant risk differences between Women and Men.")
```

 Significant Margin Differences by Gender

```python

group_women = data[data['Gender'] == 'Female']
group_men = data[data['Gender'] == 'Male']

t_stat_margin, p_margin = stats.ttest_ind(group_women['Margin'].dropna(), group_men['Margin'].dropna())
print("T-test for Margin p-value between Women and Men:", p_margin)
if p_margin < 0.05:
    print("Reject H₀: There are significant margin differences between Women and Men.")
else:
    print("Accept H₀: There are no significant margin differences between Women and Men.")
```
