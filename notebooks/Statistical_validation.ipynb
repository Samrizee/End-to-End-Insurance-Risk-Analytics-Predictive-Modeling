### Claim Frequency

```python
# Create a new column for Claim Frequency
data['ClaimFrequency'] = np.where(data['TotalClaims'] > 0, 1, 0)
```

```python
# Calculate the overall claim frequency
overall_claim_frequency = data['ClaimFrequency'].mean()
print("Overall Claim Frequency:", overall_claim_frequency)
```

## Claim Severity 

```python
# Create a column to indicate if a claim has been made
data['HasClaim'] = data['TotalClaims'] > 0

# Calculate Claim Severity only for those who made a claim
data['ClaimSeverity'] = np.where(data['HasClaim'], data['TotalClaims'] / data['TotalPremium'], 0)

# To get the average claim severity for those who made a claim
average_claim_severity = data[data['HasClaim']]['ClaimSeverity'].mean()
print("Average Claim Severity:", average_claim_severity)
```

The output of Average Claim Severity: inf indicates that there is a division by zero occurring in the calculation of ClaimSeverity. This typically happens when TotalPremium is zero for some records where TotalClaims is greater than zero.

```python
zero_premium_count = data[data['TotalPremium'] == 0].shape[0]
print("Number of records with zero TotalPremium:", zero_premium_count)
```

```python
# Create a column to indicate if a claim has been made
data['HasClaim'] = data['TotalClaims'] > 0

# Calculate Claim Severity only for those who made a claim
data['ClaimSeverity'] = np.where(data['HasClaim'] & (data['TotalPremium'] > 0),
                                  data['TotalClaims'] / data['TotalPremium'],
                                  0)

# To get the average claim severity for those who made a claim
average_claim_severity = data[data['HasClaim'] & (data['TotalPremium'] > 0)]['ClaimSeverity'].mean()
print("Average Claim Severity:", average_claim_severity)
```

### Margin

```python
data['Margin'] = data['TotalPremium'] - data['TotalClaims']
data.head()
```

### Data Segmentation

Segmentation by Province

```python
group_a = data[data['Province'] == 'Western Cape']
group_b = data[data['Province'] == 'Gauteng']
```

### Statistical Testing

Chi-Squared Test for Categorical Data

```python
# Chi-squared test for Gender across Provinces
contingency_table = pd.crosstab(data['Province'], data['Gender'])
chi2, p_gender, dof, expected = stats.chi2_contingency(contingency_table)

print("Chi-squared Test p-value for Gender:", p_gender)

# Interpret results
if p_gender < 0.05:
    print("Reject the null hypothesis: Gender has a significant impact on Province.")
else:
    print("Fail to reject the null hypothesis: Gender does not have a significant impact on Province.")
```

T-Test for Numerical Data

```python
# T-test for Margin between the two groups
t_stat_margin, p_val_margin = stats.ttest_ind(group_a['Margin'].dropna(), group_b['Margin'].dropna())
print("T-test for Margin p-value:", p_val_margin)

# Interpret results
if p_val_margin < 0.05:
    print("Reject the null hypothesis: The Province has a significant impact on Margin.")
else:
    print("Fail to reject the null hypothesis: The Province does not have a significant impact on Margin.")
```
